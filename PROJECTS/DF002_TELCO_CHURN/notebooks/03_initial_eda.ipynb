## 3. Visão Inicial do Tagert (`Churn`)

````python
contagem_churn = df.Churn.value_counts().to_frame(name="contagem")
contagem_churn["%"] = (
    contagem_churn.contagem / contagem_churn.contagem.sum() * 100
).round(2)
contagem_churn
````

Aproximadamente 1 e cada 4 clientes apresentou churn, parece que existe uma taxa alta de cancelamento no serviço. Para podermos cravar esta hipótese necessitaríamos conhecer a taxa de adesão também, mas não será nosso caso. Iremos trabalhar com os motivos que levam a esse cancelamento, quais colunas influenciam direta ou indiretamente esse resulyado. Logo é de extrema importância podermos dar um parecer técnico as lideranças.
