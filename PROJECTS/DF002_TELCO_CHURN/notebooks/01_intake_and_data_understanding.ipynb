### *Dicionário das Features Iniciais*

| Coluna | Descrição | Valores Possíveis |
|-|-|-|
| customerID | ID do cliente | Padrão da empresa |
| gender | sexo | Male: Homem; Female: Mulher |
| SeniorCitizen | Cliente é idoso? | 1: sim; 0 = não |
| Partner | Cliente possui parceiro/a? | Yes: sim; No: não |
| Dependents | Cliente tem dependentes? | Yes: sim; No: não |
| tenure | Números de meses que o cliente permaneceu na empresa | Números inteiros maiores que 0 |
| PhoneService | cliente possui serviço telefônico? | Yes: sim; No: não |
| MultipleLines | Possui múltiplas linhas telefônicas? | Yes: sim; No: não; No phone service: não possui serviço telefônico |
| InternetServices | Tipo de serviço de internet contratado pelo cliente | DSL: DSL; Fiber optic: Fibra óptica; No: Sem internet |
| OnlineSecurity | Cliente possui serviço adicional de segurança online? | Yes: sim; No: não; No internet service: Sem serviço de internet |
| OnlineBackup | Cliente possui serviço de backup online? | Yes: sim; No: não; No internet service: sem serviço de internet |
| DeviceProtection | Cliente possui proteção para dispositivos? | Yes: sim; No: não; No internet service: sem serviço de internet |
| TechSupport | Cliente possui suporte técnico? | Yes: sim; No: não; No internet service: sem serviço de internet |
| StreamingTV | Cliente possui serviço de streaming de TV | Yes: sim; No: não; No internet service: sem serviço de internet |
| StreamingMovies | Cliente possui serviço de streaming de filmes | Yes: sim; No: no; No internet service: sem serviço de internet |
| Contract | Tipo de contrato | Month-to-month: mensal; Two year: bienal; One year: anual |
| PaperlessBilling | Cliente usa faturamento sem papel (automático) | Yes: sim; No: não |
|PaymentMethod | Método de pagamento | Electronic check: cheque eletrônico; Mailed check: cheque enviado por correio; Bank transfer: transferência bancária automática; Credit card: cartão de crédito |
| MonthlyCharges | Valor cobrado mensalmente | Números de ponto flutuante maior que 0 |
| TotalCharges | Valor total cobrado | Números de ponto flutuante maior que 0 |
| Churn | Cliente cancelou o serviço? | Yes: sim; No: não |
