## 10. Cardinalidade Categórica

O dataset possui 18 colunas categóricas, muitas das quais seria interessante transformação, mas neste momento estamos apenas analisando o grau de cardinalidade. Irei adotar 3 classificação, cardinalidade 2 chamarei de binária, entre 3 e 10 baixa , maior que 10 alta e maior que 30 irei averiguar melhor (exemplo `TotalCharges` deverá possuir alto valor de cardinalidade pois trata-se de valor final da conta).

````python
colunas_categoricas
````

````python
def classificar_cardinalidade(valor):
    if valor == 2:
        return "Binária"
    elif 3 <= valor <= 10:
        return "Baixa"
    elif 11 <= valor <= 30:
        return "Moderada"
    else:
        return "Alta"
````

````python
cardinalidade = pd.DataFrame(
    {"cardinalidade": [df[col].nunique() for col in colunas_categoricas]},
    index=colunas_categoricas,
)

cardinalidade["grau"] = cardinalidade.cardinalidade.map(
    classificar_cardinalidade
)
cardinalidade["valores"] = [
    (
        ", ".join(sorted(df[col].dropna().unique().tolist()))
        if cardinalidade.loc[col, "grau"] in ["Binária", "Baixa"]
        else None
    )
    for col in colunas_categoricas
]
````
